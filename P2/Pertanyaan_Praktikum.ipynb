```python
import cv2 as cv
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
import sys # Import sys module

img = cv.imread('/content/sample_data/ktp.jpeg') # Changed path to full path

# Add a check to see if the image was loaded successfully
if img is None:
    print("Error: Could not load image from the specified path. Please verify the file integrity.")
    sys.exit("Image loading failed.") # Exit if image not loaded

print("--- Tampilan cv2_imshow ---")
cv2_imshow(img)

print("\n--- Tampilan plt.imshow ---")
plt.imshow(img)
plt.show()

# Bukti Nilai Piksel pada koordinat (y=50, x=50)
pixel_val = img[50, 50]
print("\n[BUKTI NILAI PIKSEL]")
print(f"Nilai piksel pada img[50, 50]: {pixel_val}")
print(f"Kanal 0 (OpenCV=Blue, Matplotlib diangap=Red) : {pixel_val[0]}")
print(f"Kanal 1 (OpenCV=Green, Matplotlib dianggap=Green): {pixel_val[1]}")
print(f"Kanal 2 (OpenCV=Red, Matplotlib dianggap=Blue)  : {pixel_val[2]}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

canvas_int16 = np.zeros(shape=(500, 500, 3), dtype=np.int16)
canvas_int32 = np.zeros(shape=(500, 500, 3), dtype=np.int32)

print("[Sifat Kanvas int16]")
print("dtype   :", canvas_int16.dtype)
print("min()   :", canvas_int16.min())
print("max()   :", canvas_int16.max())
print("nbytes  :", canvas_int16.nbytes, "byte")

print("\n[Sifat Kanvas int32]")
print("dtype   :", canvas_int32.dtype)
print("min()   :", canvas_int32.min())
print("max()   :", canvas_int32.max())
print("nbytes  :", canvas_int32.nbytes, "byte")

# Menampilkan kanvas
plt.subplot(1, 2, 1)
plt.imshow(canvas_int16)
plt.title('int16')

plt.subplot(1, 2, 2)
plt.imshow(canvas_int32)
plt.title('int32')
plt.show()
```
