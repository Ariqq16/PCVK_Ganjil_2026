```python
import numpy as np
import cv2 as cv

ktm = cv.imread('ktm.jpeg')
if ktm is None:
    print("Error: Could not load image 'ktm.jpeg'. Please ensure the file exists.")
else:
    mean_original = np.mean(ktm)
    h_orig, w_orig = ktm.shape[:2]
    img_small = cv.resize(ktm, (w_orig // 4, h_orig // 4), interpolation=cv.INTER_AREA)

    # Upscaling kembali ke ukuran semula
    img_restored = cv.resize(img_small, (w_orig, h_orig), interpolation=cv.INTER_CUBIC)
    mean_restored = np.mean(img_restored)

    # Menghitung selisih rata-rata intensitas
    diff_mean = abs(mean_original - mean_restored)

    print(f"Rata-rata Intensitas Asli     : {mean_original:.4f}")
    print(f"Rata-rata Intensitas Restorasi : {mean_restored:.4f}")
    print(f"Beda Rata-rata Intensitas      : {diff_mean:.4f}")
```
