```python
# ---Label Identitas Otomatis & Adaptif ---
def tambahkan_label_dokumentasi(img_path):
    img = cv.imread(gym.jpeg)
    img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    h, w = img.shape[:2]

    # tinggi bar dan ukuran teks secara proporsional
    bar_height = h // 12
    font_scale = h / 600.0
    thickness = max(1, int(h / 300))

    # bar hitam transparan/solid di bagian bawah
    bar = np.zeros((bar_height, w, 3), dtype=np.uint8)

    # bar di bawah gambar
    img_labeled = np.vstack((img, bar))

    # teks di tengah bar
    text = "Dokumentasi PCVK 2026"
    font = cv.FONT_HERSHEY_SIMPLEX
    text_size = cv.getTextSize(text, font, font_scale, thickness)[0]

    text_x = (w - text_size[0]) // 2
    text_y = h + (bar_height + text_size[1]) // 2

    cv.putText(img_labeled, text, (text_x, text_y), font, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    return img_labeled

# Pengujian dengan dua ukuran citra berbeda
# img_labeled1 = tambahkan_label_dokumentasi('foto_900x600.jpg')
# img_labeled2 = tambahkan_label_dokumentasi('foto_480x760.jpg')


# ---Frame Persegi 1:1 Adaptif Marketplace ---
def buat_thumbnail_tokopedia(img_path, nama_toko="TOKO SAYA"):
    img = cv.imread(gym.jpeg)
    img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    h, w = img.shape[:2]

    # ukuran kanvas persegi berdasarkan dimensi terbesar
    max_dim = max(h, w)

    # persegi berwarna abu-abu muda (240, 240, 240)
    canvas = np.full((max_dim, max_dim, 3), 240, dtype=np.uint8)

    # posisi offset agar citra di tengah kanvas
    offset_x = (max_dim - w) // 2
    offset_y = (max_dim - h) // 2

    # Tempatkan citra di tengah kanvas
    canvas[offset_y:offset_y + h, offset_x:offset_x + w] = img

    # logo/nama toko secara proporsional di pojok kiri atas
    font_scale = max_dim / 800.0
    thickness = max(1, int(max_dim / 400))
    margin = int(max_dim * 0.04)

    # Background banner kecil untuk nama toko
    text_size = cv.getTextSize(nama_toko, cv.FONT_HERSHEY_SIMPLEX, font_scale, thickness)[0]
    cv.rectangle(canvas, (margin, margin),
                 (margin + text_size[0] + 20, margin + text_size[1] + 20), (50, 50, 50), -1)

    cv.putText(canvas, nama_toko, (margin + 10, margin + text_size[1] + 10),
               cv.FONT_HERSHEY_SIMPLEX, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    return canvas
```
