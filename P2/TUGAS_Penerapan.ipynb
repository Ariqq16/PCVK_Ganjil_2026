```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img_bgr = cv.imread('/content/sample_data/gym.jpeg')
img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

# --Kanal Merah-Biru dan Hijau-Biru ---
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0

img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_rb)
plt.title('Kombinasi Merah - Biru')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_gb)
plt.title('Kombinasi Hijau - Biru')
plt.axis('off')
plt.tight_layout()
plt.show()

# --Pencerminan Vertikal, Horizontal, dan Kedua-duanya ---
flip_v = cv.flip(img_rgb, 0)  # Vertikal
flip_h = cv.flip(img_rgb, 1)  # Horizontal
flip_vh = cv.flip(img_rgb, -1) # Keduanya

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(flip_v)
plt.title('Pencerminan Vertikal')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_h)
plt.title('Pencerminan Horizontal')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_vh)
plt.title('Pencerminan Vertikal & Horizontal')
plt.axis('off')
plt.tight_layout()
plt.show()

img_annotated = img_rgb.copy()
h, w, _ = img_annotated.shape

cv.rectangle(img_annotated, (int(w*0.3), int(h*0.2)), (int(w*0.7), int(h*0.8)), (255, 0, 0), 3)
cv.circle(img_annotated, (int(w*0.5), int(h*0.35)), int(w*0.12), (0, 255, 0), 3)

font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(img_annotated, 'Nama: Ariq Baihaqi', (int(w*0.05), int(h*0.9)),
           font, 1.0, (255, 255, 0), 2, cv.LINE_AA)

plt.figure(figsize=(6, 6))
plt.imshow(img_annotated)
plt.title('Gambar Beranimasi / Anotasi')
plt.axis('off')
plt.show()
```
