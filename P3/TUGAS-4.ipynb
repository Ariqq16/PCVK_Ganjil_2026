```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt


img_path = '/content/drive/MyDrive/PCVK_2026/peppers.jpg'
img = cv.imread(img_path)

if img is None:
    print(f"Error: File tidak ditemukan di {img_path}")
else:
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

    R = img_rgb[:, :, 0].astype(np.float32)
    G = img_rgb[:, :, 1].astype(np.float32)
    B = img_rgb[:, :, 2].astype(np.float32)

    # a. Averaging
    gray_avg = ((R + G + B) / 3.0).astype(np.uint8)

    # b. Lightness
    max_rgb = np.maximum(np.maximum(R, G), B)
    min_rgb = np.minimum(np.minimum(R, G), B)
    gray_lightness = ((max_rgb + min_rgb) / 2.0).astype(np.uint8)

    # c. Luminance
    gray_luminance = (0.21 * R + 0.72 * G + 0.07 * B).astype(np.uint8)

    titles = ['a. Averaging', 'b. Lightness', 'c. Luminance']
    gray_images = [gray_avg, gray_lightness, gray_luminance]

    for title, gray_img in zip(titles, gray_images):
        plt.figure(figsize=(10, 4))

        # (Kiri)
        plt.subplot(1, 2, 1)
        plt.imshow(img_rgb)

        # (Kanan)
        plt.subplot(1, 2, 2)
        plt.imshow(gray_img, cmap='gray')

        plt.suptitle(title, x=0.15, y=0.98, fontsize=12, fontweight='bold')
        plt.tight_layout()
        plt.show()
```
