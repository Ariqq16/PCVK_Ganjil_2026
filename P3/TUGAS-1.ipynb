```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

# 1. Tentukan path dan baca citra peppers.jpg
img_path = '/content/drive/MyDrive/PCVK_2026/peppers.jpg'
img = cv.imread(img_path)

# Check apakah gambar berhasil dibaca
if img is None:
    print(f"Error: Could not load image from {img_path}")
else:
    # 2. Konversi dari BGR (OpenCV) ke RGB (Matplotlib)
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

    # 3. Operasi Inverse Citra: s = 255 - r
    img_inverse = 255 - img_rgb

    # 4. Tampilkan hasil citra asli dan inverse citra secara berdampingan
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Citra Asli')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(img_inverse)
    plt.title('Inverse Citra')
    plt.axis('off')


    plt.show()
```
