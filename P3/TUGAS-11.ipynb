```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

crayfish_path = '/content/drive/MyDrive/PCVK_2026/crayfish.jpg'
img_crayfish = cv.imread(crayfish_path)

if img_crayfish is not None:
    img_crayfish_rgb = cv.cvtColor(img_crayfish, cv.COLOR_BGR2RGB)

    alpha = 1.8
    beta = -40
    enhanced = np.clip(alpha * img_crayfish_rgb.astype(np.float32) + beta, 0, 255).astype(np.uint8)

    gamma_val = 0.8
    norm = enhanced / 255.0
    final_crayfish = (np.power(norm, gamma_val) * 255).astype(np.uint8)

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img_crayfish_rgb)
    plt.title('Citra Asli (Before)')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(final_crayfish)
    plt.title('Citra Hasil Perbaikan (After)')
    plt.axis('off')
    plt.show()
```
