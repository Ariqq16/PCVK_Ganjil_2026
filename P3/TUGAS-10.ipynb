```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

img_path = '/content/drive/MyDrive/PCVK_2026/houses.jpg'
img = cv.imread(img_path)

if img is not None:
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)

    gamma = 0.5
    img_normalized = img_rgb / 255.0
    gamma_corrected = np.power(img_normalized, gamma)
    result_gamma = (gamma_corrected * 255).astype(np.uint8)

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(img_rgb)
    plt.title('Citra Malam (Sebelum)')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(result_gamma)
    plt.title(f'Gamma Correction (γ = {gamma})')
    plt.axis('off')
    plt.show()
```
